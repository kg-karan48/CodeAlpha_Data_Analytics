```python
#Check Missing Value
df.isnull().sum()
```

```python
#Handling Missing Values
df["Offer"] = df["Offer"].fillna("No Offer")
df["Internet Type"] = df["Internet Type"].fillna("No Internet")
df["Churn Category"] = df["Churn Category"].fillna("Not Churned")
df["Churn Reason"] = df["Churn Reason"].fillna("Not Churned")
```

```python
df.isnull().sum().sum()
```

```python
#Checks Duplicate values
df.duplicated().sum()
```

```python
#CReating a useful columns
#Tenure GRoup
df["Tenure Group"] = pd.cut(
    df["Tenure in Months"],
    bins=[0, 12, 24, 48, float("inf")],
    labels=["0-1 Year", "1-2 Years", "2-4 Years", "4+ Years"]
)
```

```python
#Monthly Revenew Per Tenure
df["Revenue Per Month"] = df["Total Revenue"] / df["Tenure in Months"]
```

```python
#Referals Rate
df["Referral Rate"] = (
    df["Number of Referrals"] / df["Tenure in Months"].replace(0, 1)
)
```

```python
#Find Outliers
import matplotlib.pyplot as plt
sns.boxplot(x=df["Total Revenue"])
plt.show()
```

```python
sns.boxplot(x=df['Churn Score'])
plt.show()
```

```python
sns.countplot(x="Churn Label", data=df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
sns.countplot(x="Gender", hue="Churn Label", data=df)
plt.title("Churn by Gender")
plt.show()
```

```python
churn_rate = (
    df.groupby("Internet Type")["Churn Label"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

sns.barplot(
    x=churn_rate.values,
    y=churn_rate.index
)

plt.xlabel("Churn Rate (%)")
plt.ylabel("Internet Type")
plt.title("Churn Rate by Internet Type")
plt.show()
```

```python
sns.violinplot(
    data=df,
    x="Churn Label",
    y="Monthly Charge"
)
plt.title("Monthly Charge Distribution by Churn")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Tenure in Months",
    y="Total Revenue",
    hue="Churn Label",
    size="Monthly Charge"
)
plt.show()
```

```python
sns.boxplot(x="Churn Label", y="Revenue Per Month", data=df)
plt.show
```

```python
sns.histplot(df["Age"], kde=True)
plt.show()
```

```python
plt.figure(figsize=(12,8))
numeric_cols = df.select_dtypes(include="number")
sns.heatmap(numeric_cols.corr(), annot=True, cmap="coolwarm")

plt.show()
```

```python
churn_rate = df.groupby("Contract")["Churn Label"].apply(
    lambda x: (x == "Yes").mean() * 100
)

churn_rate.plot(kind="bar")

plt.ylabel("Churn Rate (%)")
plt.title("Churn Rate by Contract")
plt.show()
```

```python
pd.crosstab(
    df["Contract"],
    df["Churn Label"],
    normalize="index"
).plot(kind="bar", stacked=True)

plt.ylabel("Percentage")
plt.title("Churn Distribution by Contract")
plt.show()
```

```python
df["Churn Label"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Customer Churn Distribution")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="Monthly Charge",
    hue="Churn Label",
    kde=True
)

plt.title("Monthly Charge Distribution by Churn")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.countplot(
    data=df,
    x="Churn Label",
    ax=axes[0, 0]
)

sns.boxplot(
    data=df,
    x="Churn Label",
    y="Monthly Charge",
    ax=axes[0, 1]
)

sns.boxplot(
    data=df,
    x="Churn Label",
    y="Tenure in Months",
    ax=axes[1, 0]
)

sns.scatterplot(
    data=df,
    x="Tenure in Months",
    y="Total Revenue",
    hue="Churn Label",
    ax=axes[1, 1]
)

plt.tight_layout()
plt.show()
```
